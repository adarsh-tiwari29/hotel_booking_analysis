## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# --- STEP 1: Handling Duplicate Rows ---
print(f"Original dataset shape (with duplicates): {df.shape}")

# Removing duplicate rows to avoid skewed analysis
df = df.drop_duplicates().reset_index(drop=True)
print(f"Dataset shape after removing duplicates: {df.shape}")


# --- STEP 2: Handling Missing / Null Values properly ---
# First fill missing values securely for each column
df['children'] = df['children'].fillna(0)

if 'city' in df.columns:
    df['city'] = df['city'].fillna('Unknown')

df['agent'] = df['agent'].fillna(0)
df['company'] = df['company'].fillna(0)

# Verify if any missing values remain in these columns
print("\nRemaining missing values after cleaning:")
print(df.isnull().sum()[df.isnull().sum() > 0])


# --- STEP 3: Correcting Data Types & Cleaning Anomalies ---
# Safely converting children column to integer
df['children'] = df['children'].astype('int64')

# Removing bookings where total number of guests is zero (adults + children + babies = 0)
df['total_guests'] = df['adults'] + df['children'] + df['babies']
zero_guest_rows = df[df['total_guests'] == 0]
if not zero_guest_rows.empty:
    df = df[df['total_guests'] > 0]
    print(f"\nRemoved {len(zero_guest_rows)} bookings where total guests were zero.")

# Fixing negative or extreme anomalies in Average Daily Rate (adr)
df = df[df['adr'] >= 0]


# --- STEP 4: Feature Engineering (Creating useful columns for analysis) ---
# Total length of stay (weekend nights + week days nights)
df['total_stay'] = df['stays_in_weekend_nights'] + df['stays_in_weekdays_nights']

print("\nData Wrangling completed successfully! Dataset is now analysis-ready.")
print(f"Final clean dataset shape: {df.shape}")
````

### What all manipulations have you done and insights you found?

During the Data Wrangling phase, the following cleaning steps and manipulations were performed:
1. Removed 31,994 duplicate rows to prevent bias and ensure accurate exploratory data analysis.
2. Handled missing values strategically:
   - 'children' missing values were filled with 0, as absence of entry implies no children.
   - 'country' missing values were marked as 'Unknown'.
   - 'agent' and 'company' missing values were replaced with 0, indicating direct bookings without third-party agencies or corporate entities.
3. Performed anomaly checks: Filtered out invalid entries such as bookings with zero total guests and negative ADR (Average Daily Rate) values.
4. Feature Engineering: Created a new column 'total_stay' (sum of weekend nights and week nights) and 'total_guests' to simplify future visualizations and trend analysis.
